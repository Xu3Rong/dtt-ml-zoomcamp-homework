```python
#Q6 Median value of horsepower
# 1. Find the median value of the horsepower column in the dataset.
# 2. Next, calculate the most frequent value of the same horsepower column.
# 3. Use the fillna method to fill the missing values in the horsepower column with the most frequent value from the previous step.
# 4. Now, calculate the median value of horsepower once again.

# 1. Calculate the initial median
initial_median = df['horsepower'].median()
print("Initial median:", initial_median)

# 2. Find the most frequent value (mode)
mode_val = df['horsepower'].mode()[0]
print("Most frequent value (mode):", mode_val)

# 3. Fill missing values with the mode
hp_filled = df['horsepower'].fillna(mode_val)

# 4. Calculate the new median after imputation
new_median = hp_filled.median()
print("New median after fillna:", new_median)

# 5. Compare the values
print("-" * 30)
if new_median > initial_median:
    print("Answer: Yes, it increased")
elif new_median < initial_median:
    print("Answer: Yes, it decreased")
else:
    print("Answer: No")
```

```python
#7 Sum of weights
# 1. Select all the cars from Asia
# 2. Select only columns vehicle_weight and model_year
# 3. Select the first 7 values
# 4. Get the underlying NumPy array. Let's call it X.
# 5. Compute matrix-matrix multiplication between the transpose of X and X. To get the transpose, use X.T. Let's call the result XTX.
# 6. Invert XTX.
# 7. Create an array y with values [1100, 1300, 800, 900, 1000, 1100, 1200].
# 8. Multiply the inverse of XTX with the transpose of X, and then multiply the result by y. Call the result w.
# 9. What's the sum of all the elements of the result?

# 1. Select all cars from Asia
asia_cars = df[df['origin'] == 'Asia']

# 2. Select columns vehicle_weight and model_year
subset = asia_cars[['vehicle_weight', 'model_year']]

# 3. Select the first 7 values
subset_7 = subset.iloc[:7]

# 4. Get the underlying NumPy array X
X = subset_7.values

# 5. Compute matrix-matrix multiplication between X.T and X
XTX = X.T.dot(X)  # or X.T @ X

# 6. Invert XTX
XTX_inv = np.linalg.inv(XTX)

# 7. Create an array y
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

# 8. Multiply the inverse of XTX with X.T, then by y to get w
# Formula: w = (X^T * X)^(-1) * X^T * y
w = XTX_inv.dot(X.T).dot(y)  # or XTX_inv @ X.T @ y

# 9. Compute the sum of all elements of w
sum_w = w.sum()

print("Array w:", w)
print("Sum of elements in w:", sum_w)
```
